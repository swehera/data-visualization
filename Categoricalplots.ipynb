```python
import matplotlib.pyplot as plt 
import seaborn as sns 
```

# Categorical Data Plots
Now let's discuss using seaborn to plot categorical data! There are a few main plot types for this:

* boxplot
* violinplot
* stripplot
* swarmplot
* barplot
* countplot


Let's go through examples of each!

```python
df = sns.load_dataset("tips")
df
```

# countplot

```python
# lets see female or male counts
sns.countplot(x = df['sex'])
```

```python
# now check how many male are smoking or female smoking
sns.countplot(x = df['sex'], hue=df['smoker'])
```

# barplot

```python
sns.barplot(x = df['sex'], y = df['total_bill'])
```

```python
import numpy as np 
sns.barplot(x = df['sex'], y = df['total_bill'], estimator=np.sum)
```

```python
# now see total tips are given by female or male
sns.barplot(x = df['sex'], y = df['tip'], estimator=np.sum)
```

# box plot 

```python
sns.boxplot(x='tip', y = 'day', data=df)
```

# violinplot

```python
sns.violinplot(x = 'day', y = 'total_bill', data= df)
```

# Stripplot
Seaborn এর stripplot হলো এমন একটা plot যেখানে প্রতিটা individual data point আলাদা আলাদা dot হিসেবে দেখানো হয়।

মানে:
* প্রতিটা data value আলাদা করে দেখা যায়
* X-axis এ category থাকে
* X-axis এ category থাকে

```python
sns.stripplot(x='day', y='total_bill', data=df)
```

এখানে:

* day = category (Sun, Mon, Tue…)
* total_bill = numerical value

# Swarmplot কী?

Seaborn এর swarmplot হলো stripplot-এর upgraded version।

এটা এমনভাবে points arrange করে যাতে:

👉 কোনো point overlap না করে

```python
sns.swarmplot(x='day', y='total_bill', data=df)
```

# Stripplot vs Swarmplot (Simple Difference)

| Feature   | Stripplot                          | Swarmplot                         |
|----------|-----------------------------------|----------------------------------|
| Overlap  | হতে পারে                          | হয় না                            |
| Layout   | random spread (jitter)            | organized arrangement            |
| Speed    | fast                              | একটু slow                        |
| Best for | small data quick view             | clean visualization              |

# Combine Swarmplot and Violinplot together

```python
sns.violinplot(x='tip', y='day', data=df)
sns.swarmplot(x='tip', y='day', data=df, size=3)
```
