```python
import seaborn as sns
import plotly.express as px

tips = sns.load_dataset('tips')

px.line(tips, y='total_bill')
```

# category অনুযায়ী tip

```python
px.bar(tips, x='day', y='tip', color='sex')
```

# average tip per day (better insight)

```python
px.bar(tips.groupby('day')['tip'].mean().reset_index(),x='day', y='tip')
```

```python
px.scatter(tips, x='total_bill', y='tip')
```

```python
# gender অনুযায়ী color
px.scatter(tips, x='total_bill', y='tip', color='sex')
```

```python
# size add করলে insight বাড়ে
px.scatter(tips, x='total_bill', y='tip', size='size')
```

```python
# full advanced view
px.scatter(tips, x='total_bill', y='tip', color='day', size='size')
```

```python
px.box(tips, x='tip',y='total_bill')
```
