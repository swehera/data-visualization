```python
import seaborn as sns
import matplotlib.pylab as plt
```

***We Use distribution plots to visualize the distribution of quantitative  data***

We use distribution plots to visualize the distribution of quantitative data.


Let's discuss some plots that allow us to visualize the distribution of a data set. These plots are:

* distplot/histplot
* jointplot
* pairplot
* rugplot

```python
df = sns.load_dataset("tips")
df
```

```python
df["size"].unique()
```

```python
plt.subplot(1,2,1)
sns.histplot(df['total_bill'],bins = 20,kde = True)
plt.subplot(1,2,2)
sns.histplot(df['tip'],bins = 20,kde = True)
```

```python
fig, axes = plt.subplots(1,2, figsize=(10,4))

sns.histplot(df['total_bill'], ax=axes[0], kde=True)
sns.histplot(df['tip'], ax=axes[1], kde=True)
```

# learn Jointplot
ধরো তোমার dataset এ আছে:

* total_bill
* tip

এখন তুমি জানতে চাও:

* bill বাড়লে কি tip বাড়ে?
* bill এর distribution কেমন?
* tip এর distribution কেমন?

এই সব এক graph এ দেখার জন্য jointplot() ব্যবহার হয়।

## Basic Syntax
sns.jointplot(x='column1', y='column2', data=df)

# Example

```python
tips = sns.load_dataset("tips")

sns.jointplot(x='total_bill', y='tip', data=tips, kind="scatter")
```

```python
sns.jointplot(x='total_bill', y='tip', data=tips, kind="kde")
```

```python
sns.jointplot(x='total_bill', y='tip', data=tips, kind="hex")
```

```python
sns.jointplot(x='total_bill', y='tip', data=tips, kind="reg")
```
