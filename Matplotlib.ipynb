```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
```

***Matplotlib Basics***

```python
x = np.linspace(0,5,11)
x
```

```python
y = x ** 3
y
```

```python
plt.plot(x,y) # line graph
plt.title("Line Graph (Hira)")
plt.xlabel(" X Axis")
plt.ylabel(" Y Axis")
```

***Subplots in Matplotlib***

```python
plt.subplot(2,2,1)
plt.plot(y**2,x**2)
plt.subplot(2,2,2)
plt.plot(x,y)
plt.subplot(2,2,3)
plt.plot(y,x)
plt.subplot(2,2,4)
plt.plot(x**2,x**2)
```

***Object Oriented Way***

```python
fig = plt.figure(figsize=(10,8), dpi=100)

axis1 = fig.add_axes([0.1, 0.1, 1, 1]) # [left, bottom, width, height]
axis1.plot(x, y)
axis1.set_title("Axis One")   # ✅ correct
axis1.set_xlabel("Axis One X Axis")
axis1.set_ylabel("Axis One Y Axis")

axis2 = fig.add_axes([0.2, 0.4, .5, .5]) # [left, bottom, width, height]
axis2.plot(x, y)
axis2.set_title("Axis Two")   # ✅ correct
axis2.set_xlabel("Axis Two X Axis")
axis2.set_ylabel("Axis Two Y Axis")
```

***Saving Plots***

```python
fig = plt.figure()
ax = fig.add_axes([0.1, 0.1, 0.8, 0.8]) # [left, bottom, width, height] 
ax.plot(x,y)
plt.savefig("basic_plot.png") # save the plot in image format, create a new image file
```
